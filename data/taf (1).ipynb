```python
def skill_encoder(skill):
    return skills[skill]

#Compris, si "skills" est une variable qualitative, alors la fonction `skill_encoder` agit probablement comme un encodeur qui attribue un identifiant numérique unique à chaque compétence.


#- La fonction prend un argument `skill`, qui est le nom d'une compétence.
#- Elle utilise cette compétence comme clé pour accéder à une valeur spécifique à partir de la variable `skills`.
#- La valeur retournée est l'identifiant numérique associé à cette compétence dans la variable `skills`.

#Si `skills` est un dictionnaire où les clés sont des compétences et les valeurs sont des identifiants numériques, cette fonction récupère l'identifiant numérique correspondant à la compétence donnée. Par exemple, si `skills` est défini comme `{'Python': 1, 'Java': 2, 'SQL': 3, ...}`, alors `skill_encoder('Python')` retournera `1`, `skill_encoder('Java')` retournera `2`, etc.

#Cela peut être utile pour convertir des compétences textuelles en identifiants numériques, ce qui peut être nécessaire dans certains algorithmes d'apprentissage automatique ou d'analyse de données.
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.1, random_state = 42, stratify=y)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
# Define the hyperparameters and their possible values
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [10, 12, 15, 17, 20],
    'min_samples_split': [10, 12, 15, 17, 20],
    'max_leaf_nodes': [20, 25, 30, 35, 40],
    'min_samples_leaf': [2, 5, 7, 10],
}

# Create a Decision Tree classifier
dt_classifier = DecisionTreeClassifier(random_state=42)

# Create a GridSearchCV object
grid_search = GridSearchCV(estimator=dt_classifier, param_grid=param_grid, cv=5, scoring='accuracy')

# Fit the grid search to the data
grid_search.fit(X_train, y_train)
```

```python
# Get the best hyperparameters from the grid search
best_params = grid_search.best_params_
print("Best Hyperparameters:", best_params)
```

```python
# Get the best hyperparameters from the grid search
best_params = grid_search.best_params_
print("Best Hyperparameters:", best_params)
```

```python
# Get the best model with the best hyperparameters
DT_Model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 5, max_leaf_nodes = 6)
```

```python
DT_Model.fit(X_train, y_train)
```

```python
test_accuracy = DT_Model.score(X_test, y_test)
```

```python
print("Test Accuracy:", round(test_accuracy*100,2))
```

```python
train_sizes, train_scores, test_scores = learning_curve(
    DT_Model, X_train, y_train, cv=5, train_sizes=np.linspace(0.1, 1.0, 10), n_jobs=-1)

# Calculate the mean and standard deviation of training and test scores
train_scores_mean = np.mean(train_scores, axis=1)
train_scores_std = np.std(train_scores, axis=1)
test_scores_mean = np.mean(test_scores, axis=1)
test_scores_std = np.std(test_scores, axis=1)

# Plot the learning curve
plt.figure(figsize=(10, 6))
plt.title("Learning Curve")
plt.xlabel("Training Examples")
plt.ylabel("Score")
plt.grid()

plt.fill_between(train_sizes, train_scores_mean - train_scores_std, train_scores_mean + train_scores_std, alpha=0.1, color="r")
plt.fill_between(train_sizes, test_scores_mean - test_scores_std, test_scores_mean + test_scores_std, alpha=0.1, color="g")

plt.plot(train_sizes, train_scores_mean, 'o-', color="r", label="Training Score")
plt.plot(train_sizes, test_scores_mean, 'o-', color="g", label="Cross-validation Score")

plt.legend(loc="best")
plt.show()


#Une courbe d'apprentissage (learning curve en anglais) est un outil de diagnostic important en apprentissage automatique qui permet d'évaluer la performance d'un modèle en fonction de la taille de l'ensemble de données d'entraînement. Elle affiche généralement les performances du modèle (par exemple, l'erreur ou la précision) sur l'ensemble d'entraînement et sur l'ensemble de validation (ou de test) en fonction du nombre d'exemples d'entraînement utilisés.

#Voici quelques points clés à retenir sur les courbes d'apprentissage :

#1. **Taille de l'ensemble de données d'entraînement :** L'axe des x de la courbe d'apprentissage représente la taille de l'ensemble de données d'entraînement, souvent exprimée en nombre d'exemples.

#2. **Performance du modèle :** L'axe des y de la courbe d'apprentissage représente la performance du modèle. Cela peut être l'erreur (par exemple, l'erreur quadratique moyenne pour les tâches de régression) ou la précision (pour les tâches de classification), mais cela dépend du problème spécifique.

#3. **Train Score vs. Validation/Test Score :** Les courbes d'apprentissage comprennent généralement deux courbes : une courbe pour la performance sur l'ensemble d'entraînement (train score) et une autre pour la performance sur l'ensemble de validation (ou de test). En général, on s'attend à ce que la performance sur l'ensemble d'entraînement diminue à mesure que la taille de l'ensemble d'entraînement augmente, tandis que la performance sur l'ensemble de validation ou de test augmente ou converge vers une valeur constante.

#4. **Identification des problèmes de modèle :** Les courbes d'apprentissage peuvent aider à identifier divers problèmes de modèle, tels que le sur-ajustement (overfitting) ou le sous-ajustement (underfitting). Par exemple, un modèle surajusté montrera une grande différence entre la performance sur l'ensemble d'entraînement et sur l'ensemble de validation, tandis qu'un modèle sous-ajusté aura des performances médiocres sur les deux ensembles.

#En résumé, les courbes d'apprentissage sont un outil essentiel pour évaluer et diagnostiquer les performances des modèles d'apprentissage automatique en fonction de la taille de l'ensemble de données d'entraînement. Elles aident à comprendre comment la taille de l'ensemble de données affecte les performances du modèle et peuvent être utilisées pour améliorer la sélection du modèle et le processus d'entraînement.
```
